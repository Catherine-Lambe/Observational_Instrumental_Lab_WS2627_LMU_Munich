```python
fig = plt.figure(figsize=(10, 20))
ax = fig.subplots(10, 1)
# plt.scatter(task_5_system_stab['Zernike order - compte d\'histogramme'], task_5_system_stab['Zern Value - compte d\'histogramme'], color='blue', label='Zernike Values')
for ii in range(10):
    for jj in range(20):
        ax[ii].scatter(task_5_system_stab['Zernike order - compte d\'histogramme'][ii*20 + jj], task_5_system_stab['Zern Value - compte d\'histogramme'][ii*20 + jj], color='blue', label='Zernike Values' if ii == 0 and jj == 0 else "")
        ax[ii].xticks(np.arange(0, 21, 2))
plt.xlabel('Zernike Order')
plt.ylabel('Zernike Value')
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
# plt.scatter(task_5_system_stab['Zernike order - compte d\'histogramme'], task_5_system_stab['Zern Value - compte d\'histogramme'], color='blue', label='Zernike Values')
for ii in range(10):
    for jj in range(20):
        plt.scatter(task_5_system_stab['Zernike order - compte d\'histogramme'][ii*20 + jj]+0.05*ii, task_5_system_stab['Zern Value - compte d\'histogramme'][ii*20 + jj], alpha = 0.5, color=['blue', 'red', 'green', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan'][ii])
plt.xticks(np.arange(1, 21, 1))
plt.xlabel('Zernike Order')
plt.ylabel('Zernike Value')
plt.title('Zernike Values Stability')
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
# plt.scatter(task_5_system_stab['Zernike order - compte d\'histogramme'], task_5_system_stab['Zern Value - compte d\'histogramme'], color='blue', label='Zernike Values')
for ii in range(6):
    if ii > 0:
        yarr_err = np.array(task_5_system_stab.loc[task_5_system_stab['Zernike order - compte d\'histogramme'] == ii]['Zern Value - compte d\'histogramme'])
        plt.plot(np.arange(0, len(yarr_err))*10,np.array(task_5_system_stab.loc[task_5_system_stab['Zernike order - compte d\'histogramme'] == ii]['Zern Value - compte d\'histogramme']), alpha = 0.5)
        # plt.fill_between(x = np.arange(0, 13), y1 = yarr_err+0.005, y2 = yarr_err-0.005, alpha = 0.1)
        plt.errorbar(np.arange(0, len(yarr_err))*10,yarr_err, yerr = 0.005, alpha = 0.5, fmt = 'o', capsize = 3)
plt.xlabel('Time s*10')
plt.ylabel('Zernike Value')
plt.tight_layout()
plt.show()
```
