## Task: Transforming signals by mixing [pg.18-19 of Lab Manual]
- Try to mix the range 1416 to 1424 MHz down to 0 to 8 MHz.
- In a first step define fW , ∆f . What is the optimum fLO?
- Try to plot 2 sin(fW t) cos(fLOt). 
    - What do see in the range between: 0 and 1/∆f
    -                      and between: 0 and 1/(fW + fLO).
- Also try to shift the phase by π/10. What happens?

```python
fW = 1420
Df = 4
fLO = fW - Df
t = np.linspace(0.,1./Df,50000)
F1 = np.sin( fW *t * 2.*np.pi );
F2 = np.cos( fLO*t * 2.*np.pi );
F12 = 2.*F1*F2;
```

```python
fig = plt.figure(num=1,figsize=(10, 5),dpi=100,facecolor='w',edgecolor='k')
plt.plot(t,F12,color='blue',linewidth=1)
plt.xlabel("$t [\mu s]$")
plt.ylabel("$U [V]$")
plt.axis([0.,1./(Df),-2.,2.]) # look at the different frequency ranges
    #plt.axis([0.,1./(fW+fLO),-2.,2.])
plt.show()
```
